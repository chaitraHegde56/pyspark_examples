#### Sets

```python
set1 = {}  # This creates an empty dictionary, not a set
print(type(set1))  # Output: <class 'dict'>
set1 = set()  # This is how you create an empty set
set1 = {1, 2, 3, 4, 5, 5, 5, "abc"}
print(set1)  # Output: {1, 2, 3, 4, 5}, duplicates are removed in sets
set2 = {4, 5, 6, 7}
print(set1.union(set2))  # Output: {1, 2, 3, 4, 5, 6, 7}, all unique elements from both sets
print(set1.intersection(set2))  # Output: {4, 5}
print(set1.difference(set2))  # Output: {1, 2, 3}, elements in set1 but not in set2
print(set1.symmetric_difference(set2))  # Output: {1, 2, 3, 6, 7}, elements in either set1 or set2 but not in both
set1.add(6)  # adds 6 to set1 at the end
# set1.remove(7) # KeyError
set2.remove(7) # removes 7 from set2
set1.discard(7) # does not throw error if element not found
# print(set1[0])  # sets are unordered and do not support indexing
print("Length", set1, len(set1))  # Output: 7, number of unique elements in set1
```

```python
# map
# Other way: We perform operation on entire list using list comprehension, it is more pythonic and faster than map
li = [1, 2, 3, 4, 5]

def square1(li):
    return [x*x for x in li]

sq = square1(li)
print("Other way", sq, type(sq))
# map way: we can use map to apply a function to each element of the list, it returns an iterator, we can convert it to list or tuple or set etc.
def square(x):
    return x*x

sq = map(square, li)
print("Map way: ", list(sq), type(sq))  # Output: [1, 4, 9, 16, 25] <class 'map'>, map returns an iterator, we can convert it to list or tuple or set etc.
```

#### OOP

#### Constructor: It is a special method that is called when an object is created. It is used to initialize the attributes of the class. In python, constructor is defined using __init__ method


```python
class employee():
    company_name = "ABC Pvt Ltd"  # is a class attribute, shared by all objects of the class
    def __init__(self, name, dept):
        self.emp_name = name  # these are instance attributes which are subjected to object
        self.emp_dept = dept
    
    # Change class variable, method3: alternative to @classmethod
    def change_company_name(self, new_name):
        employee.company_name = new_name  # change class variable for all objects

    def emp_details(self):  # example for instance method
        print(f"Employee {self.emp_name} works in {self.emp_dept} for {self.company_name}")

    @staticmethod  # method which is not related to any class attribute, self is not passed. 
    # If you remove @staticmenthod, self will be passed by default and it will throw error
    def addition(a, b):
        print("Static method:", a+b)

    @classmethod
    def change_company_name_classmethod(cls, new_name):
        cls.company_name = new_name
        # Change class variable, method4: Recommended
        # cls is used to refer to class variable, it is similar to self but for class variables, it is used in class methods

emp1 = employee("Poo", "IT")
emp1.emp_details()  # Output: Employee Poo works in IT for ABC Pvt Ltd
emp2 = employee("Ravi", "HR")

# Change class variable, method1
emp2.company_name = "XYZ Pvt Ltd"  # Change class variable only for emp2
emp2.emp_details()  # Output: Employee Ravi works in HR for XYZ Pvt Ltd
emp1.emp_details()  # Output: Employee Poo works in IT for ABC Pvt Ltd

# Change class variable, method2
employee.company_name = "New Company"  # Change class variable for all objects
emp1.emp_details()  # Output: Employee Poo works in IT for New Company
emp2.emp_details()  # Output: Employee Ravi works in HR for XYZ Pvt Ltd, class variable is not changed for emp2 because it was overridden earlier

emp1.change_company_name("Another Company")  # Change class variable for all objects using method
emp1.emp_details()  # Output: Employee Poo works in IT for Another Company

emp1.addition(5, 10)  # Output: 15, static method can be called using object

emp1.change_company_name_classmethod("Class_Method Company")
emp1.emp_details()  # Output: Employee Poo works in IT for Class_Method Company, class variable is changed for all objects
```

#### type of methods:
##### Instance method: It is a method that takes self as first parameter and can access instance attributes and class attributes. It can modify the state of the object. It can be called using object name.
##### Class method: It is a method that takes cls as first parameter and can access class attributes. It can modify the state of the class. It can be called using class name or object name.
##### Static method: It is a method that does not take self or cls as first parameter and cannot access instance attributes or class attributes. It is used for utility functions. It can be called using class name or object name.


#### Getters and Setters

```python
class employees():
    def __init__(self, name, dept):
        self.emp_name = name
        self.emp_dept = dept

    @property  # this is getter
    def display(self):
        print(f"Employee {self.emp_name} works in {self.emp_dept}")

    @display.setter  # this is setter
    def display(self, details_list):
        self.emp_name = details_list[0]
        self.emp_dept = details_list[1]

emp1 = employees("Poo", "IT")
emp1.display = ["Poooo", "HR"]  # setter
emp1.display  # getter, Output: Employee Poooo works in HR
```

#### Inheritance

```python
# Single level
class company():
    def __init__(self, com_name):
        self.company_name = com_name

    def display_company(self):
        print(f"Company name {self.company_name}")

class employee(company):
    def __init__(self, emp_name, com_name):
        self.emp_name = emp_name
        self.company_name = com_name

    def display_emp(self):
        print(f"Emp name {self.emp_name} company_name {self.company_name}")

emp1 = employee("Rahul", "ABC")
emp1.display_emp()
emp1.display_company()
```

```python
class company():
    def __init__(self, com_name, country):
        self.company_name = com_name
        self.country = country

    def display_company(self):
        print("Inside parent")
        print(f"Company name {self.company_name} country {self.country}")

class employee(company):
    def __init__(self, emp_name, com_name, country):
        self.emp_name = emp_name

        # Method1 to initialize parent attribute
        company.__init__(self, com_name, country)
        
        # Method2 to initialize parent attribute
        # self.company_name = com_name
        # self.country = country
        # Method 1 is useful when we have many parent attributes

    def display_emp(self):
        print(f"Emp name {self.emp_name} company name {self.company_name} country {self.country}")

    def display_company_child(self):
        print("Inside child")
        company.display_company(self)
        # OR use super(): refers ONLY ONE LEVEL UP, useful when we have only single level inheritance, mess when we have multi level
        super().display_company()

emp1 = employee("Rahul", "ABC", "India")
emp1.display_company_child()
emp1.display_emp()
```

#### Multiple inheritance
##### child_class inherits from parent_class1 and parent_class2

```python
class company():  # Parent1
    def __init__(self, company_name):
        self.company_name = company_name

    def display_company(self):
        print(f"Company name {self.company_name}")

class region():
    def __init__(self, region_name):
        self.region_name = region_name

    def display_region(self):
        print(f"Region name {self.region_name}")

class employee(company, region):  # Child
    def __init__(self, emp_name, company_name, region_name):
        self.emp_name = emp_name
        company.__init__(self, company_name)
        region.__init__(self, region_name)

    def display_emp(self):
        print(f"Employee {self.emp_name} works for company {self.company_name} in region {self.region_name}")

    def child_info(self):
        company.display_company(self)
        region.display_region(self)
        super().display_company()
        super().display_region()

emp1 = employee("Rahul", "ABC", "Asia")
emp1.display_emp()
emp1.child_info()
```

#### Multi level inheritance

```python
class company():
    def __init__(self, com_name):
        self.company_name = com_name

    def display_company(self):
        print(f"Company name {self.company_name}")

    def showme(self):
        print("I am grandfather")

class department(company):
    def __init__(self, dept_name, com_name):
        self.dept_name = dept_name
        company.__init__(self, com_name)

    def display_department(self):
        print(f"Department name {self.dept_name} country {self.company_name}")

    # def showme(self):
    #     print("I am father")

class employee(department):
    def __init__(self, emp_name, dept_name, com_name):
        self.emp_name = emp_name
        department.__init__(self, dept_name, com_name)

    def display_emp(self):
        print(f"Employee {self.emp_name} works at department {self.dept_name} at company {self.company_name}")

    def display_all(self):
        company.display_company(self)
        department.display_department(self)

    # def showme(self):
    #     print("I am pro")

emp1 = employee("Rahul", "IT", "Amazon")
emp1.display_emp()
emp1.display_department()  # Same as department.display_department(self) in display_all
emp1.display_company()  # Same as company.display_company(self) in display_all
emp1.display_all()
emp1.showme()
```

#### Multithreading

```python
import time
import random
from concurrent.futures import ThreadPoolExecutor

tables = ["tab1", "tab2", "tab3", "tab4", "tab5"]

def my_func(i):
    # wait = random.randint(1, 10)
    wait = 3
    time.sleep(wait)
    return f"{i} took {wait} seconds"

start_time = time.perf_counter()

for i in tables:
    result = my_func(i)
    print(result)

end_time = time.perf_counter()
elapsed_time = end_time - start_time
print(f"Total time {elapsed_time:.4f} seconds")
```

```python
# Method 1: using map
start_time = time.perf_counter()
with ThreadPoolExecutor(max_workers=len(tables)) as executor:
    futures = executor.map(my_func, tables)  # Futures is a var which is holding returned value by my_func
print(f"Final result type {futures}")
print(f"Final result {list(futures)}")
end_time = time.perf_counter()
elapsed_time = end_time - start_time
print(f"Total time {elapsed_time:.4f} seconds")
```

```python
# Method 2: using submit
start_time = time.perf_counter()
with ThreadPoolExecutor(max_workers=len(tables)) as executor:
    for i in tables:
        futures = executor.submit(my_func, i)
end_time = time.perf_counter()
elapsed_time = end_time - start_time
print(f"Total time {elapsed_time:.4f} seconds")
```
